```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler,LabelEncoder
fifa=pd.read_csv("../data/fifa_world_cup_2026_player_performance.csv")
print("FIFA dataset loaded successfully.")
print("Shape:",fifa.shape)
print("Columns:",len(fifa.columns))
```

```python
print("Columns:"); print(fifa.columns.tolist()); print("\nFirst five records:"); display(fifa.head())
```

```python
missing_values=fifa.isnull().sum(); print("Missing-value count by column:"); display(missing_values.to_frame("Missing Values")); print("Total missing values:",int(missing_values.sum()))
```

```python
fifa_clean=fifa.copy(); numeric_cols=fifa_clean.select_dtypes(include=np.number).columns; categorical_cols=fifa_clean.select_dtypes(include=["object","category"]).columns
for col in numeric_cols:
    if fifa_clean[col].isnull().any(): fifa_clean[col]=fifa_clean[col].fillna(fifa_clean[col].mean())
for col in categorical_cols:
    if fifa_clean[col].isnull().any():
        mode_value=fifa_clean[col].mode()
        if not mode_value.empty: fifa_clean[col]=fifa_clean[col].fillna(mode_value.iloc[0])
print("Total missing values before:",int(fifa.isnull().sum().sum())); print("Total missing values after:",int(fifa_clean.isnull().sum().sum()))
```

```python
comparison=pd.DataFrame({"Before Cleaning":fifa.isnull().sum(),"After Cleaning":fifa_clean.isnull().sum()}); display(comparison)
```

```python
duplicate_count=int(fifa_clean.duplicated().sum()); print("Duplicate rows before removal:",duplicate_count); fifa_clean=fifa_clean.drop_duplicates().copy(); print("Duplicate rows after removal:",int(fifa_clean.duplicated().sum())); print("Cleaned dataset shape:",fifa_clean.shape)
```

```python
outlier_col="market_value_eur"; q1=fifa[outlier_col].quantile(.25); q3=fifa[outlier_col].quantile(.75); iqr=q3-q1; lower=q1-1.5*iqr; upper=q3+1.5*iqr; outliers=fifa[(fifa[outlier_col]<lower)|(fifa[outlier_col]>upper)]
print("Column selected:",outlier_col); print("Q1:",q1); print("Q3:",q3); print("IQR:",iqr); print("Lower limit:",lower); print("Upper limit:",upper); print("Number of outliers:",len(outliers)); display(outliers.head(10))
```

```python
plt.figure(figsize=(8,4)); plt.boxplot(fifa[outlier_col].dropna()); plt.title("Market Value – Outlier Detection"); plt.ylabel("market_value_eur"); plt.show()
```

```python
scaler=MinMaxScaler(); fifa_transformed=fifa_clean.copy(); fifa_transformed["market_value_normalized"]=scaler.fit_transform(fifa_transformed[["market_value_eur"]]); display(fifa_transformed[["market_value_eur","market_value_normalized"]].head(10)); print("Minimum normalized value:",fifa_transformed["market_value_normalized"].min()); print("Maximum normalized value:",fifa_transformed["market_value_normalized"].max())
```

```python
encoder=LabelEncoder(); fifa_transformed["team_encoded"]=encoder.fit_transform(fifa_transformed["team"].astype(str)); print("Column encoded: team"); print("Number of categories:",len(encoder.classes_)); print("First five category mappings:"); print(dict(zip(encoder.classes_[:5],encoder.transform(encoder.classes_[:5])))); display(fifa_transformed[["team","team_encoded"]].drop_duplicates().sort_values("team_encoded").head(10))
```

```python
one_hot=pd.get_dummies(fifa_clean[["position"]].astype(str),prefix="position"); print("One-hot encoded shape:",one_hot.shape); display(one_hot.head(10))
```

```python
print("Preprocessing completed successfully.")
print("Final cleaned shape:",fifa_clean.shape)
```

```python
import pandas as pd
df=pd.read_csv("../../data/fifa_world_cup_2026_player_performance.csv")
print("First 5 records of FIFA dataset:")
display(df.head())
print("Dataset Shape:",df.shape)
print("Number of Columns:",len(df.columns))
```

```python
import matplotlib.pyplot as plt
team_goals=df.groupby("team")["goals"].sum().sort_values(ascending=False).head(15)
plt.figure(figsize=(10,5)); plt.plot(team_goals.index,team_goals.values,marker="o"); plt.title("Top 15 FIFA Teams by Total Player Goals"); plt.xlabel("Team"); plt.ylabel("Total Goals"); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

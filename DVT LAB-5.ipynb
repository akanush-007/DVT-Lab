```python
import matplotlib.pyplot as plt
```

```python
teams = ["Qatar", "Netherlands", "Panama", "Cameroon", "Saudi Arabia", "Jamaica"]
goals = [95, 94, 90, 88, 82, 79]
```

```python
plt.plot(teams, goals, marker="o", color="blue", linewidth=2)
plt.title("Top Teams Goal Trend in FIFA 2026 World Cup")
plt.xlabel("Team")
plt.ylabel("Total Goals Scored")

plt.grid(True)
plt.show()
```

```python
plt.bar(teams, goals, color="skyblue")
plt.title("Team Goals Comparison")
plt.xlabel("Team")
plt.ylabel("Goals")
plt.show()
```

```python
positions = ["Defender", "Midfielder", "Forward", "Goalkeeper"]
player_count = [16380, 16380, 16380, 5460]

plt.pie(player_count, labels=positions, autopct="%1.1f%%", startangle=140)
plt.title("Player Position Distribution")
plt.show()
```

```python
teams = ["Qatar", "Netherlands", "Panama", "Cameroon"]
goals = [95, 94, 90, 88]
assists = [70, 68, 65, 61]

plt.plot(teams, goals, label="Goals", marker="o")
plt.plot(teams, assists, label="Assists", marker="s")
plt.title("Goals vs Assists Comparison")
plt.xlabel("Team")
plt.ylabel("Count")
plt.legend()
plt.show()
```

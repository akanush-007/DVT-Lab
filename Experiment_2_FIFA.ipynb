```python
import pandas as pd
import sqlite3
fifa=pd.read_csv("../data/fifa_world_cup_2026_player_performance.csv")
print("FIFA CSV file imported successfully.")
print("Rows:",len(fifa))
print("Columns:",len(fifa.columns))
```

```python
display(fifa.head())
```

```python
fifa.info()
```

```python
display(fifa.describe(include="all").T)
```

```python
fifa_sample=fifa.head(20); fifa_sample.to_excel("fifa_sample.xlsx",index=False)
fifa_excel=pd.read_excel("fifa_sample.xlsx")
print("Excel file imported successfully."); display(fifa_excel.head())
```

```python
conn=sqlite3.connect("fifa_data.db"); fifa_sample.to_sql("fifa_players",conn,if_exists="replace",index=False)
fifa_sql=pd.read_sql("SELECT * FROM fifa_players LIMIT 5",conn)
print("SQL data imported successfully."); display(fifa_sql); conn.close()
```

```python
print("Web import syntax demonstrated using pd.read_csv(url).")
print("Example: url = \"https://example.com/fifa_data.csv\"; web_data = pd.read_csv(url)")
```

```python
fifa_sample.to_csv("fifa_export.csv",index=False); print("FIFA CSV file exported successfully."); print("Exported rows:",len(fifa_sample))
```

```python
fifa_sample.to_excel("fifa_export.xlsx",index=False); print("FIFA Excel file exported successfully."); print("Exported rows:",len(fifa_sample))
```

```python
fifa_sample.to_json("fifa_export.json",orient="records",indent=2); print("FIFA JSON file exported successfully."); print("Exported records:",len(fifa_sample))
```

```python
conn=sqlite3.connect("fifa_export.db"); fifa_sample.to_sql("fifa_data",conn,if_exists="replace",index=False); count=pd.read_sql("SELECT COUNT(*) AS row_count FROM fifa_data",conn); print("FIFA data exported to SQLite successfully."); display(count); conn.close()
```

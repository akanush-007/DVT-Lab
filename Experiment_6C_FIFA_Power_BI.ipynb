```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("../../data/fifa_world_cup_2026_player_performance.csv")
print("Dataset loaded successfully")
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
df.head()
```

```python
print("Duplicate rows:",df.duplicated().sum()); print("\nMissing values:"); display(df.isna().sum().to_frame("Missing Values"))
```

```python
df["match_date"]=pd.to_datetime(df["match_date"],errors="coerce"); df["goal_involvement"]=df["goals"]+df["assists"]; df["is_clean_sheet"]=df["clean_sheet"].astype(int); print("Transformed columns added: goal_involvement, is_clean_sheet"); print("Date range:",df.match_date.min().date(),"to",df.match_date.max().date()); print("Total goals:",int(df.goals.sum()))
```

```python
metrics=pd.DataFrame({"Metric":["Total Matches","Total Player Records","Total Goals","Total Assists","Teams"],"Value":[df.match_id.nunique(),len(df),int(df.goals.sum()),int(df.assists.sum()),df.team.nunique()]}); metrics
```

```python
stage_goals=df.groupby("tournament_stage")["goals"].sum(); plt.figure(figsize=(11,5)); stage_goals.plot(kind="line",marker="o"); plt.title("FIFA Total Goals by Tournament Stage"); plt.xlabel("Tournament Stage"); plt.ylabel("Goals"); plt.grid(True,alpha=.3); plt.tight_layout(); plt.show()
```

```python
team_goals=df.groupby("team")["goals"].sum().sort_values(ascending=False).head(10); plt.figure(figsize=(10,6)); team_goals.sort_values().plot(kind="barh"); plt.title("Top 10 FIFA Teams by Total Goals"); plt.xlabel("Goals"); plt.ylabel("Team"); plt.tight_layout(); plt.show()
```

```python
position_goals=df.groupby("position")["goals"].sum(); plt.figure(figsize=(8,8)); plt.pie(position_goals.values,labels=position_goals.index,autopct="%1.1f%%",startangle=90); plt.title("Goal Contribution by Position"); plt.tight_layout(); plt.show()
```

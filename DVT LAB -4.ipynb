```python
import pandas as pd

df = pd.read_csv("fifa_world_cup_2026_player_performance.csv")

print(df[["player_name", "team", "position", "goals", "player_rating"]].head())
```

```python

print("Dataset Shape:", df.shape)
```

```python

print(df[["player_name", "team", "goals", "minutes_played", "player_rating"]].dtypes)
```

```python

high_rating_players = df[df["player_rating"] > 8.5]

print(high_rating_players[["player_name", "team", "position", "player_rating"]].head())
```

```python

spain_players = df[df["team"] == "Spain"]

print(spain_players[["player_name", "position", "goals", "assists", "player_rating"]].head())
```

```python

sorted_players = df.sort_values(by="player_rating", ascending=False)

print(sorted_players[["player_name", "team", "goals", "player_rating"]].head(10))
```

```python

print("Mean =", df["player_rating"].mean())
print("Median =", df["player_rating"].median())
print("Mode =", df["player_rating"].mode()[0])
print("Standard Deviation =", df["player_rating"].std())
```

```python

print(df[["goals", "assists", "minutes_played", "player_rating"]].describe())
```

```python

team_avg_rating = df.groupby("team")["player_rating"].mean().reset_index()

print(team_avg_rating.head(10))
```

```python

corr = df[["goals", "minutes_played", "player_rating"]].corr()

print("Correlation Matrix:")
print(corr)
```

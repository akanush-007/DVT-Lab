```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("../../data/fifa_world_cup_2026_player_performance.csv")
print("Dataset loaded successfully")
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
df.head()
```

```python
df["match_date"]=pd.to_datetime(df["match_date"],errors="coerce"); df["goal_involvement"]=df["goals"]+df["assists"]; print("Dataset loaded successfully"); print("Shape:",df.shape); display(df.head())
```

```python
print("Duplicate rows:",df.duplicated().sum()); print("Date range:",df.match_date.min().date(),"to",df.match_date.max().date()); summary=pd.DataFrame({"Metric":["Matches","Teams","Player Records","Total Goals","Goal Involvement"],"Value":[df.match_id.nunique(),df.team.nunique(),len(df),int(df.goals.sum()),int(df.goal_involvement.sum())]}); display(summary)
```

```python
team_goals=df.groupby("team")["goals"].sum().sort_values(ascending=False).head(10); plt.figure(figsize=(10,6)); team_goals.sort_values().plot(kind="barh"); plt.title("Top 10 FIFA Teams by Total Goals"); plt.xlabel("Goals"); plt.ylabel("Team"); plt.tight_layout(); plt.show()
```

```python
stage_goals=df.groupby("tournament_stage")["goals"].sum(); plt.figure(figsize=(11,5)); stage_goals.plot(kind="line",marker="o"); plt.title("FIFA Goals by Tournament Stage"); plt.xlabel("Tournament Stage"); plt.ylabel("Goals"); plt.grid(True,alpha=.3); plt.tight_layout(); plt.show()
```

```python
position_goals=df.groupby("position")["goals"].sum(); plt.figure(figsize=(8,8)); plt.pie(position_goals.values,labels=position_goals.index,autopct="%1.1f%%",startangle=90); plt.title("Goal Distribution by Position"); plt.show()
```

```python
team_summary=df.groupby("team").agg(Total_Goals=("goals","sum"),Total_Assists=("assists","sum"),Avg_Rating=("player_rating","mean"),Player_Records=("player_id","size")).sort_values("Total_Goals",ascending=False).head(10); display(team_summary)
```

```python
print("Total Goals = SUM([goals])"); print("Goal Involvement = SUM([goals]) + SUM([assists])"); print("Matches = COUNTD([match_id])"); print("Players = COUNTD([player_id])")
```

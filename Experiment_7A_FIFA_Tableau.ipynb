```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv("../../data/fifa_world_cup_2026_player_performance.csv")
print("Dataset loaded successfully")
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
df.head()
```

```python
team_goals=df.groupby("team")["goals"].sum().sort_values(ascending=False); stage_goals=df.groupby("tournament_stage")["goals"].sum(); position_goals=df.groupby("position")["goals"].sum(); print("Calculated fields:"); print("Total Goals = SUM([goals])"); print("Goal Involvement = SUM([goals]) + SUM([assists])"); print("Minimum Goals parameter can be applied to filter teams.")
```

```python
plt.figure(figsize=(10,6)); team_goals.head(10).sort_values().plot(kind="barh"); plt.title("Top 10 FIFA Teams by Total Goals"); plt.xlabel("Goals"); plt.ylabel("Team"); plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(11,5)); stage_goals.plot(kind="line",marker="o"); plt.title("FIFA Goals by Tournament Stage"); plt.xlabel("Tournament Stage"); plt.ylabel("Goals"); plt.grid(True,alpha=.3); plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(8,8)); plt.pie(position_goals.values,labels=position_goals.index,autopct="%1.1f%%",startangle=90); plt.title("Goal Distribution by Position"); plt.tight_layout(); plt.show()
```

```python
minimum_goals=50; filtered_teams=team_goals[team_goals>=minimum_goals].sort_values(ascending=False); print(f"Minimum Goals parameter = {minimum_goals:,}"); print("Teams meeting the parameter:"); display(filtered_teams.to_frame("Total Goals"))
```
